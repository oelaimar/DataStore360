```python
import pandas as pd
import numpy as np
import hashlib


print("Pandas: ", pd.__version__)
print("Numpy: ", np.__version__)

# hashing
hashing = lambda x : hashlib.sha256(x.encode()).hexdigest()


# import the cleaning data
pd.set_option("display.max_columns", 25)
df_cleaned = pd.read_csv("./processed/data_cleaned.csv")

df_cleaned.columns.tolist()
```

```python

personal_columns = [
    'Customer Name',
    'Customer ID',
]

for col in personal_columns:
    if col in df_cleaned.columns:
        print(f"\n--- {col} ---")
        print(df_cleaned[col].head())
        print("Unique:", df_cleaned[col].nunique())
        print("Missing:", df_cleaned[col].isna().sum())
```

```python
# pseudonymize the Customer ID

df_cleaned['Customer ID Pseudo'] = df_cleaned['Customer ID'].astype(str).apply(hashing)

display(df_cleaned[['Customer ID', 'Customer ID Pseudo']].head())
# drop the Customer ID
df_cleaned = df_cleaned.drop(columns=['Customer ID'])
df_cleaned
```

```python
# pseudonymize the Customer Name

df_cleaned['Customer Name Pseudo'] = df_cleaned['Customer Name'].astype(str).apply(hashing)

display(df_cleaned[['Customer Name', 'Customer Name Pseudo']].head())
# drop the Customer Name
df_cleaned = df_cleaned.drop(columns=['Customer Name'])
```
